```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("tanishqdublish/urban-traffic-density-in-cities")

print("Path to dataset files:", path)
```

```python
import os

print(os.listdir(path))
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(os.path.join(path, 'futuristic_city_traffic.csv'))
```

```python
#Xem dữ liệu cơ bản
print(df.info())
```

```python
#Bài 1:In dòng đầu tiên
print(df.head(20))
```

```python
#Bài 2:Xem tên các cột
print(df.columns)
```

```python
#Bài 3:Đếm số lượng dữ liệu theo thành phố
df_city=df['City'].value_counts()
print(df_city)
```

```python
#Bài 4:Đếm số lượng dữ liệu theo phương tiện
df_vehicle = df['Vehicle Type'].value_counts()
print(df_vehicle)
```

```python
#Bài 5:Tính tốc độ trung bình của các phương tiện theo từng thành phố
df_speed = df.groupby('City')['Speed'].mean()
print(df_speed)
```

```python
#Bài 6:tính tốc độ trung bình của các phương tiện theo từng thành phố và sắp xếp kết quả từ cao xuống thấp.
df_speed_sorted = df.groupby('City')['Speed'].mean().sort_values(ascending=False)
print(df_speed_sorted)
```

```python
#Bài 7:Tính mật độ giao thông trung bình của các phương tiện theo từng thành phố
df_density = df.groupby('City')['Traffic Density'].mean()
print(df_density)
```

```python
#Bài 8:Traffic Density trung bình của từng thành phố, sau đó sắp xếp từ cao xuống thấp.
df_density_sorted = df.groupby('City')['Traffic Density'].mean().sort_values(ascending=False)
print(df_density_sorted)
```

```python
'''
Tạo thêm một cột mới tên là Speed_Class để phân loại phương tiện dựa trên tốc độ:
'''
df['Speed_Class'] = df['Speed'].apply(lambda x: 'Fast' if x >= 80 else 'Slow')
df_speed_class = df['Speed_Class'].value_counts()
print(df[['City','Speed', 'Speed_Class']])
```

```python
#Bài 10:Tạo thêm một cột mới tên Traffic_Level dựa trên Traffic Density:
df['Traffic_Level'] = df['Traffic Density'].apply(lambda x: 'High' if x >= 0.5 else 'Low')
df_traffic_level = df['Traffic_Level'].value_counts()
print(df[['City','Traffic Density', 'Traffic_Level']])
```

```python
#Bài 11:đếm số lượng phương tiện của từng loại xe trong từng thành phố.
df_vehicle_city = df.groupby(['City', 'Vehicle Type']).size().unstack(fill_value=0)
print(df_vehicle_city)
```

```python
#Bài 12: Tính tốc độ trung bình theo thành phố và loại xe
df_speed_city = df.groupby(['City', 'Vehicle Type'])['Speed'].mean()
print(df_speed_city)    
```

```python
#Bài 13:Tìm phương tiện có tốc độ trung bình cao nhất trong từng thành phố.
df_max_speed_vehicle = df.groupby(['City', 'Vehicle Type'])['Speed'].mean()
print(df_max_speed_vehicle.groupby('City').idxmax())
```

```python
#Bài 14:Tính tốc độ trung bình theo thời tiết và thành phố
df_speed_weather_city = df.groupby(['City', 'Weather'])['Speed'].mean()
print(df_speed_weather_city)
```

```python
#Bài 15:Tìm thời tiết và thời gian có tốc độ trung bình cao nhất trong từng thành phố.
df_max_speed_weather_time = df.groupby(['City', 'Weather', 'Hour Of Day'])['Speed'].mean()
print(df_max_speed_weather_time.groupby('City').idxmax())
```

```python
#Bài 16:Tìm ngày trong tuần  có mật độ giao thông trung bình cao nhất trong từng thành phố.
df_day_max_density=df.groupby(['City', 'Day Of Week'])['Traffic Density'].mean()
print(df_day_max_density.groupby('City').idxmax())
```

```python
#Bài 17:Tìm giờ có mật độ giao thông cao nhất ở từng thành phố
df_hour_max_density = df.groupby(['City', 'Hour Of Day'])['Traffic Density'].mean()
print(df_hour_max_density.groupby('City').idxmax())
```

```python
#Bài 18:Tìm thời tiết có mức tiêu thụ năng lượng cao nhất
df_weather_maxenerfy=df.groupby('Weather')['Energy Consumption'].mean()
print(df_weather_maxenerfy.idxmax())
```

```python
#Bài 19:Tìm loại phương tiện (Vehicle Type) có mức tiêu thụ năng lượng trung bình cao nhất trong từng thành phố (City).
df_vehicle_max_energy = df.groupby(['City', 'Vehicle Type'])['Energy Consumption'].mean()
print(df_vehicle_max_energy.groupby('City').idxmax())
```

```python
#Bài 20:Tìm loại phương tiện (Vehicle Type) có tốc độ trung bình cao nhất trong toàn bộ dataset.
df_vehicle_max_speed = df.groupby('Vehicle Type')['Speed'].mean()
print(df_vehicle_max_speed.idxmax())
```
