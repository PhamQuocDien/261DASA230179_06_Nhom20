```python
import pandas as pd
import numpy as np
import kagglehub
pd.set_option('display.max_rows', 2000)
```

```python

# Tải dataset
path = kagglehub.dataset_download("albertovidalrod/gapminder-dataset")
df = pd.read_csv(path + "/gapminder_data_graphs.csv")
```

```python

# 1. Quốc gia có GDP lớn nhất mỗi châu lục năm 2010
print("GDP lớn nhất theo từng châu lục năm 2010:")
df_nam=df[df["year"] == 2010]
df_lonNhat=df_nam.groupby("continent")["gdp"].idxmax().reset_index()['gdp']
print(list(df_lonNhat))


print(df_nam.loc[list(df_lonNhat),["continent", "country", "gdp"]])
```

```python
df_2018 = df[df['year'] == 2018].copy()
df_2018['phanloai'] = df_2018.apply(
    lambda x: 'Phat trien' if x['life_exp'] >= 73 and x['gdp'] >= 1800 else 'Dang phat trien',
    axis=1
)
print(df_2018[['continent','country', 'life_exp', 'gdp', 'phanloai']])
```

```python
if 'phanloai' not in df_2018.columns:
    df_2018['phanloai'] = df_2018.apply(
        lambda x: 'Phat trien' if x['life_exp'] >= 73 and x['gdp'] >= 1800 else 'Dang phat trien',
        axis=1
    )

a = (df_2018['phanloai'] == 'Dang phat trien').sum()
print(f"Số lượng quốc gia đang phát triển năm 2018: {a}")
b=(df_2018['phanloai']=='Phat trien').sum()
print(f"Số lượng quốc gia phát triển năm 2018: {b}")
```

```python
df_phattrien=df_2018.groupby('continent')['phanloai'].value_counts().unstack().fillna(0)
print(df_phattrien)
```
